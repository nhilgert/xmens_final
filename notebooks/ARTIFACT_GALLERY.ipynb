Scripts and documents are listed with their `host.artifact_path(...)` call so you can open them.

```
import json, pandas as pd
from IPython.display import Image, display, JSON
pd.set_option('display.max_columns', 40)

def show_img(vid, name):
    print(name); display(Image(filename=host.artifact_path(vid)))

def show_tab(vid, name, n=8):
    try:
        df = pd.read_csv(host.artifact_path(vid), sep=None, engine='python')
        print(f'{name}   shape={df.shape}'); display(df.head(n))
    except Exception as e:
        print(f'{name}: {e}')

def show_json(vid, name):
    print(name); display(JSON(json.load(open(host.artifact_path(vid)))))
```
