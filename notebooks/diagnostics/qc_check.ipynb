```python
import anndata as ad
import os
import pandas as pd

s = pd.read_csv('/n/eddy_lab/users/nhilgert/xmens_final/config/samples.tsv', sep='\t')
tot = {'mouse': 0, 'human': 0}

for _, r in s.iterrows():
    f = f"data/interim/qc/{r['sample_id']}.h5ad"
    if os.path.exists(f):
        n = ad.read_h5ad(f, backed='r').n_obs
        tot[r['species']] += n
        print(f"{r['sample_id']:24s} {r['species']:6s} {n}")
    else:
        print(f"{r['sample_id']:24s} MISSING")

print(f"\nMOUSE: {tot['mouse']}  (paper initial-QC 68146, diff {tot['mouse']-68146:+d})")
print(f"HUMAN: {tot['human']}  (paper initial-QC 70250, diff {tot['human']-70250:+d})")
```
