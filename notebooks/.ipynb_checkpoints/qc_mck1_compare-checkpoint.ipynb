# Single-cell QC

Per-sample quality control for post-CellBender `.h5ad` files: computes QC metrics, drops low-quality barcodes, removes doublets, and writes cleaned `.h5ad`s.

**Before running:** fill in the configuration cell (`base_path`, `samples`, `output_dir`) and review the QC thresholds for your dataset.

**Input contract** — each sample lives at `{base_path}/{sample}/{input_filename}` and must have:
- raw (un-normalized) counts in `.X` (or point `counts_layer` at the right layer)
- gene symbols as `var_names` (used to flag mt / ribo / hb genes)

All outputs (figures + cleaned h5ads) are written under `output_dir`; nothing is written back to `base_path`.

**Requires:** `scanpy`, `anndata`, `doubletdetection` (+ `leidenalg`), `scipy`, `seaborn`, `matplotlib`.

## Configuration

```python
# ---- paths ----
# Load the PIPELINE's ingested object so this comparison is apples-to-apples
# with the snakemake qc rule (both start from the same 3938-cell ingest output).
# NOTE: the pipeline writes one .h5ad PER SAMPLE at data/interim/raw/<sample>.h5ad
# (not <sample>/<filename>), so we load by full path below rather than base_path+folder.
base_path = "data/interim/raw"                     # pipeline raw/ingest outputs
input_filename = None                              # not used (we load full paths)
output_dir = "./qc_output_mck1_compare"            # all results + figures go here

# ---- samples (explicit list) ----
# Pipeline sample_id for mck_1:
samples = [
    "1_dpi_early_mck1",
]

# ---- counts ----
# The pipeline ingest stores corrected counts in .X AND in layers['counts'].
# Use the counts layer so metrics + doublet detection read the same source the
# pipeline uses.
counts_layer = "counts"

# ---- QC thresholds (review per dataset) ----
min_counts   = 500    # hard floor: drop barcodes below this many total counts
max_mito_pct = 8.0    # fixed upper bound on % mitochondrial counts
nmads        = 3      # MAD multiplier for total_counts / n_genes_by_counts outliers

# ---- doublet detection ----
n_iters = 40
n_jobs  = 8           # set to your available cores
```

## Imports & setup

```python
import os
import math
import numpy as np
import pandas as pd
from scipy.stats import median_abs_deviation

import matplotlib.pyplot as plt
import seaborn as sns

import anndata as ad
import scanpy as sc
import doubletdetection

# scanpy settings
sc.settings.verbosity = 3
sc.settings.set_figure_params(dpi=150)
sc.logging.print_header()

# seaborn style
sns.set_context("notebook", font_scale=1)
sns.set_style("ticks")

blues  = sns.color_palette("Blues", 9)
reds   = sns.color_palette("Reds", 9)
greens = sns.color_palette("Greens", 9)
grays  = sns.color_palette("Greys", 9)
```

```python
# Output layout (created if missing; nothing is written back to base_path)
fig_dir         = os.path.join(output_dir, "figures")
doublet_fig_dir = os.path.join(fig_dir, "doublet_convergence")
results_dir     = os.path.join(output_dir, "h5ad")

for d in (output_dir, fig_dir, doublet_fig_dir, results_dir):
    os.makedirs(d, exist_ok=True)

print(f"Writing outputs under: {os.path.abspath(output_dir)}")
```

## Load data

```python
if not samples:
    raise ValueError("`samples` is empty \u2014 list your sample names in the config cell.")

adatas = []
for sample in samples:
    # Pipeline layout: one file per sample at base_path/<sample>.h5ad
    adata_path = os.path.join(base_path, f"{sample}.h5ad")
    if not os.path.exists(adata_path):
        raise FileNotFoundError(f"No file for sample '{sample}': {adata_path}")

    adata = sc.read_h5ad(adata_path)

    # Promote the counts layer to .X so everything downstream (QC metrics +
    # doublet detection) reads from one source, matching the pipeline.
    if counts_layer is not None:
        if counts_layer not in adata.layers:
            raise KeyError(
                f"'{sample}': counts_layer '{counts_layer}' not in layers "
                f"(available: {list(adata.layers.keys())})"
            )
        adata.X = adata.layers[counts_layer].copy()

    adata.obs["sample"] = sample
    print(f"Sample {sample}: loaded {adata.shape[0]} barcodes, {adata.shape[1]} genes")
    adatas.append(adata)

print(f"\nLoaded {len(adatas)} samples.")
```

## Compute QC metrics

Replicates the upstream metadata convention: flag mt / ribo / hb genes, then run scanpy's `calculate_qc_metrics` on the raw counts in `.X`. This produces `total_counts`, `n_genes_by_counts`, `pct_counts_mt`, `pct_counts_ribo`, `pct_counts_hb`, and the `log1p_*` / `pct_counts_in_top_20_genes` companions.

Note the metric is named **`pct_counts_mt`** (after the `mt` var flag), not `pct_counts_mito`. Filtering below uses `pct_counts_mt`, `total_counts`, and `n_genes_by_counts`; `ribo` / `hb` are computed and available but not filtered on.

```python
def add_qc_metrics(adata):
    """Flag gene classes and compute scanpy QC metrics on the counts in .X."""
    adata.var["mt"] = adata.var_names.str.startswith(("mt-", "MT-"))
    adata.var["ribo"] = adata.var_names.str.startswith((
        "rps", "rpl", "mrps", "mrpl",
        "Rps", "Rpl", "Mrps", "Mrpl",
        "RPS", "RPL", "MRPS", "MRPL",
    ))
    adata.var["hb"] = (
        adata.var_names.str.contains("^Hb[^(p)]")
        | adata.var_names.str.contains("^HB[^(P)]")
    )

    sc.pp.calculate_qc_metrics(
        adata,
        qc_vars=["mt", "ribo", "hb"],
        inplace=True,
        percent_top=[20],
        log1p=True,
    )
    return adata


for i, adata in enumerate(adatas):
    adatas[i] = add_qc_metrics(adata)

# Confirm the metrics the filtering step needs now exist.
required = ["total_counts", "n_genes_by_counts", "pct_counts_mt"]
for sample, adata in zip(samples, adatas):
    missing = [m for m in required if m not in adata.obs]
    if missing:
        raise KeyError(f"'{sample}': QC metrics missing after computation: {missing}")

print("QC metrics computed for all samples.")
```

## Helper functions

`is_outlier` flags cells more than `nmads` MADs from the median; for the heavy-tailed count metrics it works on a log10 scale and stores the bounds back on the linear scale. `plot_metric_grid` lays out one histogram per sample on a grid sized to the sample count (so it works for any number of samples, not a fixed 4×4).

```python
def log1p(x):
    return np.log10(x + 1)

def inv_log1p(y):
    return np.power(10, y) - 1

# Metrics that are MAD-thresholded on a log10 scale.
LOG_METRICS = ("total_counts", "n_genes_by_counts")


def is_outlier(adata, metric, nmads):
    """Flag cells where `metric` is more than `nmads` MADs from the median."""
    M = adata.obs[metric]
    if metric in LOG_METRICS:
        M = log1p(M)

    median_val = np.nanmedian(M)
    mad_val = median_abs_deviation(M, nan_policy="omit")
    if mad_val == 0:
        print(f"  Warning: MAD is zero for {metric}; bounds collapse to the median.")

    lower_bound = median_val - nmads * mad_val
    upper_bound = median_val + nmads * mad_val
    outlier = (M < lower_bound) | (M > upper_bound)

    if metric in LOG_METRICS:
        lower_bound = inv_log1p(lower_bound)
        upper_bound = inv_log1p(upper_bound)

    adata.uns[f"{metric}_min"] = lower_bound
    adata.uns[f"{metric}_max"] = upper_bound
    return outlier


def make_grid(n):
    """Return (fig, flat_axes) sized to n panels, capped at 4 columns."""
    ncols = min(4, n)
    nrows = math.ceil(n / ncols)
    fig, axs = plt.subplots(nrows, ncols, figsize=(4 * ncols, 4 * nrows), squeeze=False)
    axs = axs.flatten()
    for ax in axs[n:]:
        ax.axis("off")
    return fig, axs


def plot_metric_grid(metric, log_scale=False, save_name=None):
    """Per-sample histogram of `metric` with threshold lines drawn from adata.uns."""
    fig, axs = make_grid(len(adatas))
    for i, (sample, adata) in enumerate(zip(samples, adatas)):
        ax = axs[i]
        sns.histplot(adata.obs[metric], bins=50, color=blues[6],
                     log_scale=log_scale, kde=True, ax=ax)
        for key in (f"{metric}_min", f"{metric}_max"):
            bound = adata.uns.get(key)
            if bound is not None:
                ax.axvline(x=bound, color=reds[5], linestyle="--")
        ax.set_title(sample, fontsize=14)
        ax.set_xlabel(metric.replace("_", " ").title())
        ax.set_ylabel("# barcodes")
        ax.grid(visible=True, which="both", linestyle="--", linewidth=0.5)
    fig.tight_layout()
    if save_name:
        fig.savefig(os.path.join(fig_dir, save_name), dpi=300, bbox_inches="tight")
    plt.show()


def filter_outliers(adata, metric):
    keep = ~adata.obs[f"{metric}_outlier"]
    n_removed = int((~keep).sum())
    print(f"  {metric}: removing {n_removed} outliers")
    return adata[keep].copy()
```

## Filtering

Three stages: (1) a hard floor on total counts, (2) flag outliers — a fixed upper bound on mito %, and MAD-based bounds on `total_counts` / `n_genes_by_counts`, (3) apply the flags.

```python
# Stage 1: hard floor on total counts.
total_before = total_removed = 0
for i, (sample, adata) in enumerate(zip(samples, adatas)):
    n0 = adata.n_obs
    n_low = int((adata.obs["total_counts"] < min_counts).sum())
    sc.pp.filter_cells(adata, min_counts=min_counts)
    adatas[i] = adata
    total_before += n0
    total_removed += n_low
    print(f"{sample}: removed {n_low}/{n0} barcodes below {min_counts} counts ({n_low / n0:.0%})")

print(f"\nTotal below floor: {total_removed}/{total_before} ({total_removed / total_before:.0%})")
```

```python
# Stage 2a: mitochondrial % — fixed upper threshold.
for adata in adatas:
    adata.uns["pct_counts_mt_min"] = 0.0
    adata.uns["pct_counts_mt_max"] = max_mito_pct
    adata.obs["pct_counts_mt_outlier"] = adata.obs["pct_counts_mt"] > max_mito_pct

plot_metric_grid("pct_counts_mt", log_scale=False, save_name="mito_pct.png")
```

```python
# Stage 2b: total_counts and n_genes_by_counts — MAD-based.
for metric in ("total_counts", "n_genes_by_counts"):
    for adata in adatas:
        adata.obs[f"{metric}_outlier"] = is_outlier(adata, metric, nmads)
    plot_metric_grid(metric, log_scale=True, save_name=f"{metric}.png")
```

```python
# Stage 3: apply the outlier flags.
filter_metrics = ["pct_counts_mt", "total_counts", "n_genes_by_counts"]

for i, (sample, adata) in enumerate(zip(samples, adatas)):
    n0 = adata.n_obs
    print(f"{sample}: {n0} cells before outlier filtering")
    for metric in filter_metrics:
        adata = filter_outliers(adata, metric)
    adatas[i] = adata
    print(f"  -> {adata.n_obs} remain ({(n0 - adata.n_obs) / n0:.0%} removed)\n")
```

## Doublet detection

```python
for i, (sample, adata) in enumerate(zip(samples, adatas)):
    print(f"\n*** doubletdetection: {sample} ({i + 1}/{len(adatas)}) ***")

    clf = doubletdetection.BoostClassifier(
        n_iters=n_iters,
        clustering_algorithm="leiden",
        standard_scaling=True,
        n_top_var_genes=3000,
        n_jobs=n_jobs,
    )

    preds = clf.fit(adata.X).predict(p_thresh=1e-7, voter_thresh=0.5)

    doubletdetection.plot.convergence(
        clf, show=True, p_thresh=1e-7, voter_thresh=0.5,
        save=os.path.join(doublet_fig_dir, f"{sample}_doublet_convergence.pdf"),
    )

    adata.obs["doubletdetection_predicted_doublet"] = preds.astype(bool)
    adata.obs["doubletdetection_doublet_score"] = clf.doublet_score()
    adatas[i] = adata
```

## Remove doublets & save

```python
for i, (sample, adata) in enumerate(zip(samples, adatas)):
    n0 = adata.n_obs
    adata = adata[~adata.obs["doubletdetection_predicted_doublet"]].copy()
    n_removed = n0 - adata.n_obs
    print(f"{sample}: removed {n_removed} doublets ({n_removed / n0:.0%}), {adata.n_obs} remain")
    adatas[i] = adata

    outpath = os.path.join(results_dir, f"{sample}_cellbender_qc_applied.h5ad")
    adata.write_h5ad(outpath)
    print(f"  saved -> {outpath}")
```
