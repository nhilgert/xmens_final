```python
import scanpy as sc, anndata as ad, numpy as np, pandas as pd
import matplotlib.pyplot as plt

# --- load the old human object ---
H = "/n/lab_storage/eddy_lab/users/nhilgert/decidualization/revisions/adatas/submission/20251008_Cevrim_Human_Only_Fib_Dec_Cells.h5ad"   # or the deposited Cevrim_Human path
```

```python
h = ad.read_h5ad(H)
print("shape:", h.shape, "| obsm:", list(h.obsm))


# --- ensure scoring on log1p ---
if float(h.X.max()) > 50:
    if   "log1p"   in h.layers: h.X = h.layers["log1p"].copy();   print("scored on layers['log1p']")
    elif "lognorm" in h.layers: h.X = h.layers["lognorm"].copy(); print("scored on layers['lognorm']")
    else: h.layers["counts_backup"]=h.X.copy(); sc.pp.normalize_total(h,target_sum=1e4); sc.pp.log1p(h); print("normalized+log1p (X was counts)")
else:
    print("X already log-normalized")

# --- uppercase-map mouse modules -> human, REPORT hit rate per module ---
print("\northolog (naive uppercase) hit rates:")
MODULES_HS = {}
for name, genes in MODULES.items():
    hs = [g.upper() for g in genes]
    present = [g for g in hs if g in h.var_names]
    MODULES_HS[name] = present
    print(f"  {name}: {len(present)}/{len(genes)} genes  {'⚠️ SPARSE' if len(present) < len(genes)*0.4 else ''}")

# --- score each module on human ---
for name, present in MODULES_HS.items():
    if len(present) == 0:
        print(f"  {name}: 0 genes — skipping (no orthologs found)"); continue
    sc.tl.score_genes(h, present, score_name=f"module_{name}", use_raw=False)
score_cols = [f"module_{n}" for n in MODULES_HS if len(MODULES_HS[n])>0]
```

```python
# --- coordinates are in obs, not obsm, for this human object ---
XCOL, YCOL = "x_um_dbscan", "y_um_dbscan"
assert XCOL in h.obs.columns and YCOL in h.obs.columns, \
    f"coord cols missing; obs has: {[c for c in h.obs.columns if 'um' in c.lower() or 'x' in c.lower()[:2]]}"
XY = h.obs[[XCOL, YCOL]].to_numpy(dtype=float)
print(f"coords from obs['{XCOL}','{YCOL}'] | valid: {int(np.isfinite(XY).all(1).sum())}/{len(XY)}")

# --- confirm the timepoint facet column ---
print("time-ish cols:", [c for c in h.obs.columns if "time" in c.lower() or "dpi" in c.lower()])
FACET = "timepoint"   # <-- set to the actual column name from the print above
assert FACET in h.obs.columns, f"'{FACET}' not in obs; available: {list(h.obs.columns)[:40]}"
groups = sorted(h.obs[FACET].astype(str).unique())
print(f"faceting by {FACET}: {groups}")
```

```python
def plot_module_spatial(mod):
    col = f"module_{mod}"
    vals = h.obs[col].values.astype(float)
    fin = np.isfinite(vals)
    vlo, vhi = np.percentile(vals[fin], (2, 98))          # shared color scale across timepoints
    n = len(groups)
    fig, ax = plt.subplots(1, n, figsize=(4.6*n, 4.6), squeeze=False)
    for j, g in enumerate(groups):
        sel = (h.obs[FACET].astype(str) == g).values & np.isfinite(XY[:,0]) & np.isfinite(XY[:,1])
        sctr = ax[0,j].scatter(XY[sel,0], XY[sel,1], c=vals[sel], s=6, cmap="viridis",
                               vmin=vlo, vmax=vhi, linewidths=0)
        ax[0,j].set_title(f"{g}  (n={int(sel.sum())})", fontsize=9)
        ax[0,j].set_aspect("equal"); ax[0,j].set_xticks([]); ax[0,j].set_yticks([]); ax[0,j].invert_yaxis()
    fig.suptitle(f"{mod} module (human, {len(MODULES_HS[mod])}/{len(MODULES[mod])} orthologs)", fontsize=10)
    plt.colorbar(sctr, ax=ax[0,:], shrink=0.7, label="module score")
    plt.tight_layout(); plt.show()

# loop all modules with orthologs
for mod in [n for n in MODULES_HS if len(MODULES_HS[n]) > 0]:
    plot_module_spatial(mod)
```
