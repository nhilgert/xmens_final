## AUCell scoring helper

AUCell is rank-based: it ranks genes within each cell and measures enrichment of each module near
the top. Robust to depth/normalization. **Input: raw counts** (we normalize to CP10k; no log needed —
ranks are preserved). Adds `auc_<MODULE>` columns to `.obs`.

```python
def score_aucell(adata, modules, counts_layer, num_workers=8):
    """Add auc_<module> obs columns via pyscenic AUCell. Requires integer counts in `counts_layer`."""
    assert counts_layer in adata.layers, f"no layer {counts_layer!r}; have {list(adata.layers)}"
    m = adata.copy()
    m.X = m.layers[counts_layer].copy()
    # sanity: counts should be non-negative integers
    chk = m.X[:50].toarray() if hasattr(m.X, "toarray") else np.asarray(m.X[:50])
    if not np.allclose(chk, np.round(chk)):
        print("WARNING: counts layer is not integer-valued — AUCell expects raw counts.")
    sc.pp.normalize_total(m, target_sum=1e4)   # CP10k, no log
    X = m.X.toarray() if hasattr(m.X, "toarray") else np.asarray(m.X)
    ex = pd.DataFrame(X, index=m.obs_names, columns=m.var_names)

    sigs = []
    for name, genes in modules.items():
        present = [g for g in genes if g in ex.columns]
        miss = [g for g in genes if g not in ex.columns]
        if miss: print(f"  {name}: {len(miss)} genes absent (dropped)")
        sigs.append(GeneSignature(name=name, gene2weight=present))

    thr = derive_auc_threshold(ex)[0.01]      # 1st percentile of detection, per pyscenic example
    print("auc_threshold:", round(float(thr), 4))
    auc = aucell(ex, sigs, auc_threshold=thr, num_workers=num_workers)
    auc = auc.reindex(adata.obs_names)
    for name in modules:
        if name in auc.columns:
            adata.obs[f"auc_{name}"] = auc[name].values
    print("added:", [c for c in adata.obs.columns if c.startswith("auc_")])
    return adata
```

## 1. Heatmap — module AUCell scores × fibroblast/decidual subtypes

```python
import anndata as ad
a = ad.read_h5ad(HEATMAP_OBJ)
print(a.shape)
# score AUCell if not already present
if not any(c.startswith("auc_") for c in a.obs.columns):
    a = score_aucell(a, MODULES, COUNTS_LAYER)

# subtypes present, in layering order
present_order = [s for s in SUBTYPE_ORDER if s in set(a.obs["cell_subtype"].astype(str))]
missing = [s for s in SUBTYPE_ORDER if s not in present_order]
if missing: print("subtypes not in object:", missing)
sub = a[a.obs["cell_subtype"].isin(present_order)].copy()
sub.obs["cell_subtype"] = pd.Categorical(sub.obs["cell_subtype"], categories=present_order, ordered=True)

# mean AUC per subtype, min-max per module (column)
score_cols = [f"auc_{n}" for n in PANEL_ORDER]
mat = sub.obs.groupby("cell_subtype", observed=True)[score_cols].mean()
mat_mm = (mat - mat.min()) / (mat.max() - mat.min())

fig, ax = plt.subplots(figsize=(len(score_cols)*0.75+3, len(present_order)*0.55+2), facecolor="black")
ax.set_facecolor("black")
im = ax.imshow(mat_mm.values, cmap="magma", vmin=0, vmax=1, aspect="auto")
ax.set_xticks(range(len(score_cols))); ax.set_xticklabels(PANEL_ORDER, rotation=45, ha="right", color="white")
ax.set_yticks(range(len(present_order))); ax.set_yticklabels(present_order, color="white")
ax.tick_params(colors="white")
for sp in ax.spines.values(): sp.set_visible(False)
cbar = plt.colorbar(im, ticks=[0,1], pad=0.02)
cbar.set_label("AUCell score\n(min–max per module)", color="white", fontsize=14)
cbar.ax.set_yticklabels(["0","1"]); cbar.ax.tick_params(colors="white"); cbar.outline.set_edgecolor("0.3")
ax.set_title("module AUCell × fibroblast/decidual subtypes (mouse)", color="white", pad=10)
plt.savefig("aucell_heatmap.png", dpi=300, bbox_inches="tight", facecolor="black")
plt.tight_layout(); plt.show()
```

## 2. Spatial maps (3dpi)

Steps: select the 3dpi sample → remove spatial outliers (kNN k=5, one-sided 3-MAD on mean neighbor
distance) → AUCell → per-module spatial panels (min–max per module, PDZ featured, dark/magma).

```python
from sklearn.neighbors import NearestNeighbors

sp = ad.read_h5ad(SPATIAL_OBJ)
if TIMEPOINT_VALUE is not None and TIMEPOINT_COL in sp.obs.columns:
    sp = sp[sp.obs[TIMEPOINT_COL].astype(str) == TIMEPOINT_VALUE].copy()
    print(f"selected {TIMEPOINT_COL}=={TIMEPOINT_VALUE}: {sp.n_obs} cells")

# --- locate coordinates ---
def get_xy(adata):
    if "spatial" in adata.obsm: return np.asarray(adata.obsm["spatial"], float)
    for xc, yc in [("x_um_dbscan","y_um_dbscan"), ("x_um","y_um")]:
        if {xc, yc}.issubset(adata.obs.columns):
            return adata.obs[[xc, yc]].to_numpy(float)
    raise ValueError(f"no coords; obsm={list(adata.obsm)}")
XY_raw = get_xy(sp)

# --- spatial outlier removal: kNN mean-neighbor-distance, one-sided 3-MAD ---
finite = np.isfinite(XY_raw).all(1)
nn = NearestNeighbors(n_neighbors=6).fit(XY_raw[finite])   # k=5 + self
d,_ = nn.kneighbors(XY_raw[finite]); mnd = d[:,1:].mean(1)
med = np.median(mnd); mad = np.median(np.abs(mnd-med))
keep_f = mnd <= med + 3*1.4826*mad
keep = np.zeros(sp.n_obs, bool); keep[np.where(finite)[0][keep_f]] = True
print(f"spatial outlier removal: keep {keep.sum()}/{sp.n_obs}")
sub_clean = sp[keep].copy()

# --- AUCell on the cleaned 3dpi cells ---
if not any(c.startswith("auc_") for c in sub_clean.obs.columns):
    sub_clean = score_aucell(sub_clean, MODULES, COUNTS_LAYER)
```

```python
# --- spatial plot: PDZ featured left, 2x5 others, dark/magma, min-max per module, sorted draw ---
XY0 = get_xy(sub_clean)
XYf = XY0.copy(); cx, cy = np.nanmean(XYf[:,0]), np.nanmean(XYf[:,1])
if FLIP_X: XYf[:,0] = 2*cx - XYf[:,0]
if FLIP_Y: XYf[:,1] = 2*cy - XYf[:,1]
th = np.deg2rad(THETA_DEG); R = np.array([[np.cos(th),-np.sin(th)],[np.sin(th),np.cos(th)]])
XY = (XYf - [cx,cy]) @ R.T + [cx,cy]

FEATURE = PANEL_ORDER[0]
def draw(ax, mod, ts=20):
    ax.set_facecolor("black")
    ax.scatter(XY[:,0], XY[:,1], s=7, c=BASE_COLOR, linewidths=0, zorder=1)
    v = sub_clean.obs[f"auc_{mod}"].values.astype(float)
    lo, hi = np.percentile(v[np.isfinite(v)], (2,98))
    vn = np.clip((v-lo)/(hi-lo+1e-12), 0, 1); o = np.argsort(vn)
    s = ax.scatter(XY[o,0], XY[o,1], c=vn[o], s=7, cmap=CMAP, vmin=0, vmax=1, linewidths=0, zorder=2)
    ax.set_title(mod, fontsize=ts, color="white", pad=8); ax.set_aspect("equal")
    ax.set_xticks([]); ax.set_yticks([])
    for sp_ in ax.spines.values(): sp_.set_visible(False)
    return s

others = [m for m in PANEL_ORDER if m != FEATURE]
fig = plt.figure(figsize=(3.6*6, 3.9*2), facecolor="black")
gs = GridSpec(2, 6, figure=fig, wspace=0.05, hspace=0.15)
sctr = draw(fig.add_subplot(gs[:,0]), FEATURE, 24)
for i, mod in enumerate(others):
    r, c = divmod(i, 5); sctr = draw(fig.add_subplot(gs[r, c+1]), mod)
cb = fig.colorbar(sctr, ax=fig.axes, shrink=0.5, ticks=[0,1], pad=0.02)
cb.set_label("AUCell score\n(min–max per module)", color="white", fontsize=18)
cb.ax.set_yticklabels(["0","1"]); cb.ax.tick_params(colors="white", labelsize=14); cb.outline.set_edgecolor("0.3")
plt.savefig("aucell_spatial_3dpi.png", dpi=300, bbox_inches="tight", facecolor="black")
plt.show()
```

## Methods note

Module activity was scored with **AUCell** (Aibar et al., 2017), which ranks genes within each cell and
computes the area under the recovery curve for each module's gene set over the top-ranked fraction of
genes (threshold from the 1st percentile of detected genes per cell). AUCell is rank-based and therefore
robust to sequencing depth and normalization. Heatmaps show mean AUCell score per subtype, min–max
normalized per module; spatial panels show per-cell AUCell scores, min–max normalized per module, after
removal of spatial outliers (kNN k=5; cells > 3 scaled-MAD above the median mean-neighbor distance).
Module signals were concordant across alternative scoring methods (see companion notebook).
